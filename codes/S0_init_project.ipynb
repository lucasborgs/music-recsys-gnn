# S0 — Project Initialization

Creates the directory tree expected by the pipeline (data, graphs and reports
folders). It reads the folder layout from `conf/config.yaml` and **does not**
overwrite that file — the configuration is version-controlled.


```python
import os
from pathlib import Path
import yaml


def find_project_root(anchor_file="conf/config.yaml"):
    """Walk up from the current working directory until the folder containing
    `anchor_file` is found, and return it as the project root."""
    current_path = Path.cwd()
    for parent in [current_path] + list(current_path.parents):
        if (parent / anchor_file).exists():
            return parent
    raise FileNotFoundError(
        f"Could not locate the project root containing '{anchor_file}'."
    )

BASE_DIR = find_project_root()
print(f"Project root: {BASE_DIR}")

# Load the layout from the version-controlled configuration
with open(BASE_DIR / "conf/config.yaml", "r") as f:
    config = yaml.safe_load(f)

# All folders declared under `paths`, plus the graph sub-folders used downstream
folders = list(config["paths"].values()) + [
    "graphs/item_item",
    "graphs/super_item_item",
    "graphs/bipartite/coarsened",
]

for rel_path in folders:
    target = BASE_DIR / rel_path
    target.mkdir(parents=True, exist_ok=True)
    print(f"  ready: {rel_path}")

print("\nProject directory structure is ready.")
```
